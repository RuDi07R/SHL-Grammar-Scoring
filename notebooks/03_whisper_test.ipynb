# 3. Whisper transcription test

## Objective

The recordings must first be converted into text because the linguistic feature pipeline operates on transcripts. This notebook checks that Whisper can load, transcribe a representative training recording, and run with the available PyTorch/CUDA configuration.

The transcription call uses `fp16=False`, which keeps inference compatible with the selected execution path even when CUDA is available. The timing cell provides a practical estimate of inference cost without changing the transcription methodology.

```python
import os
import pandas as pd
import whisper

train = pd.read_csv("../train.csv")

filename = train.iloc[0]["filename"]
audio_path = os.path.join("../train", filename)

print("Transcribing:", filename)
print("Audio path:", audio_path)
```

```python
import torch
import whisper

print("PyTorch:", torch.__version__)
print("CUDA:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
model = whisper.load_model("base")

print("Whisper loaded!")
```

```python
model = model.to("cuda")

print("Whisper moved to GPU!")
```

```python
filename = train.iloc[0]["filename"]
audio_path = os.path.join("../train", filename)

print("Transcribing:", filename)

result = model.transcribe(
    audio_path,
    fp16=False
)

print(result["text"])

transcript = result["text"]

print("\nTRANSCRIPT:")
print(transcript)
```

```python
import time

start = time.time()

result = model.transcribe(
    audio_path,
    fp16=False
)

elapsed = time.time() - start

print(f"Time taken: {elapsed:.2f} seconds")
print(f"Audio duration: {result['segments'][-1]['end']:.2f} seconds")
print(result["text"])
```

## Interpretation

The successful model load, representative transcript, and timing output show that Whisper can provide the text input needed by the linguistic pipeline. The timing is recorded for operational context; it is not a model-quality metric. The same base Whisper configuration is reused by the training and test transcription scripts.
