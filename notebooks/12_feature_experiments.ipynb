## Experiment protocol

The experiment table is produced by the reusable script in `scripts/run_experiments.py`. It compares Ridge regularization values across acoustic, original linguistic, V2 linguistic, and combined feature sets, then evaluates conservative nonlinear models on the strongest Ridge feature sets.

All reported scores come from shuffled 5-fold cross-validation with preprocessing fitted inside each pipeline. Selection prioritizes lower RMSE and then higher Pearson correlation; the saved `selected_model.json` records the resulting final choice.

# 12. Feature experiments
Exact protocol: KFold(n_splits=5, shuffle=True, random_state=42). Preprocessing is fitted inside each sklearn Pipeline. Test data is never used.

```python
from pathlib import Path
import sys
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
results = pd.read_csv(ROOT / 'outputs' / 'experiment_results.csv')
results.sort_values(['rmse','pearson'], ascending=[True,False]).head(15)
```
