# 2. Mean-prediction baseline

## Objective

This notebook establishes a simple reference point for grammar-score prediction. Every sample is assigned the mean training label, so the result measures the error that must be improved upon by learned audio and transcript features.

RMSE measures average prediction error in score units. Pearson correlation measures whether predictions preserve the ordering of the target. The baseline is useful because it gives every later model a common, reproducible point of comparison.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import mean_squared_error
from scipy.stats import pearsonr

train = pd.read_csv("../train.csv")

X = train.drop(columns=["label"])
y = train["label"]

print("Samples:", len(y))
print("Mean score:", y.mean())
```

```python
pred = np.full(len(y), y.mean())

rmse = np.sqrt(mean_squared_error(y, pred))
pearson = pearsonr(y, pred)[0]

print("Baseline RMSE:", rmse)
print("Baseline Pearson:", pearson)
```

## Interpretation

The recorded mean baseline has RMSE `1.2381936808002134`. Pearson is `NaN` because a constant prediction has no variation. Any useful learned model must reduce RMSE substantially while producing predictions that track the ordering of grammar scores.
