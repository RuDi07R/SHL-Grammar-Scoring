# 7. Text embeddings

## Objective

Sentence embeddings represent transcript meaning and usage patterns in a dense numeric vector. The notebook uses the actually configured `SentenceTransformer("all-MiniLM-L6-v2")` model and caches one embedding per training transcript.

The embedding has 384 dimensions. This representation may capture semantic and contextual information that is not explicitly represented by counts, ratios, and punctuation features. The next notebook tests whether that additional representation improves regression performance.

```python
import pandas as pd
import numpy as np

from sentence_transformers import SentenceTransformer

PROJECT_ROOT = r"C:\Users\rudra\Desktop\SHL-Grammar-Scoring"

transcripts = pd.read_csv(
    PROJECT_ROOT + r"\outputs\train_transcripts.csv"
)

transcripts["transcript"] = transcripts["transcript"].fillna("")

print("Samples:", len(transcripts))
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded")
```

```python
texts = transcripts["transcript"].tolist()

embeddings = model.encode(
    texts,
    batch_size=32,
    show_progress_bar=True
)

print("Embedding shape:", embeddings.shape)
```

```python
embedding_path = PROJECT_ROOT + r"\outputs\text_embeddings.npy"

np.save(embedding_path, embeddings)

print("Saved:", embedding_path)
```

## Interpretation

The embedding model produces one 384-value vector per transcript and writes the cached array used by notebook 08. The embedding itself is a representation, not a score; its usefulness is established only by the downstream cross-validated regression result.
